```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("C:/Users/dasar/Downloads/Synthetic.csv")
data.head()
```

```python
data.drop(columns = ["Unnamed: 0"],inplace =True)
data.head()
```

```python
x = data.iloc[:,0].values
y = data.iloc[:,1].values
```

```python
x = (x-np.mean(x))/ np.std(x)
```

```python
theta_0 = np.array(np.random.randn()).reshape(1,1)
theta_1 = np.array(np.random.randn()).reshape(1,1)

learning_rate = 0.0001
epochs = 1000
tolerance = 0.001

m = len(x)
losses=[]
```

```python
for epoch in range(epochs):
    indices = np.random.permutation(m)
    x = x[indices]
    y = y[indices]

    for i in range(m):
        xi = x[i]
        yi = y[i]

        y_hat = theta_0*xi + theta_1
        error = y_hat - yi

        d_theta_0 = xi*error
        d_theta_1 = error

        theta_0 = theta_0 - learning_rate*d_theta_0
        theta_1 = theta_1 - learning_rate*d_theta_1

    y_hat_all = theta_0*x + theta_1
    loss = 1/2 * np.mean((y_hat_all - y)**2)

    losses.append(loss)

    if loss<= tolerance:
        break
```

```python
plt.plot(losses)
plt.xlabel("Number of epochs")
plt.ylabel("Loss MSE")
```
