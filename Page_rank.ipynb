```python
import numpy as np

nodes = ['A','B','C','D']

adj_matrix = np.array([
    [0,1,1,0],
    [0,0,1,0],
    [1,0,0,0],
    [0,0,1,0]
])
print(adj_matrix)
```

```python
N = len(nodes)
d = 0.85
tolerance = 0.001
max_iterations = 10

PR = np.ones(N) / N

outgoing_links = np.sum(adj_matrix, axis=1)

print("Initial Page Rank:")
for i in range(N):
    print(nodes[i], PR[i])

for iteration in range(max_iterations):

    new_PR = np.ones(N) * ((1-d)/N)

    for i in range(N):
        for j in range(N):
            if adj_matrix[j,i] == 1:
                new_PR[i] = new_PR[i] + d * (PR[j] / outgoing_links[j])

    print("\nIteration", iteration + 1)

    for i in range(N):
        print(nodes[i], round(new_PR[i], 6))

    if np.max(np.abs(new_PR - PR)) < tolerance:
        PR = new_PR
        print("\nConvergence reached.")
        break

    PR = new_PR
```

```python
print("\nFinal Page Rank:")
for i in range(N):
    print(nodes[i], round(PR[i], 6))
```
